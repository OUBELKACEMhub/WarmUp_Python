## Exercice 4 : Aggrégation et GroupBy

**Objectif :**
À partir du DataFrame `df1`, calculez par **Ville** :
1. Le salaire moyen.
2. L'âge maximum.

*(Indice : utilisez `.groupby()` avec `.agg()`)*

```python
import pandas as pd
import numpy as np 

data2 = {
    'Nom': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank'],
    'Age': [25, 30, 35, 40, 22, 28],
    'Ville': ['Paris', 'Lyon', 'Marseille', 'Paris', 'Lyon', 'Lille'],
    'Salaire': [35000, 42000, 50000, 48000, 29000, 38000]
}

# VOTRE CODE ICI
df6= pd.DataFrame(data2)

r=df6.groupby("Ville")["Salaire"].mean()
r1=df6.groupby("Ville")["Age"].max()
r2=df6.groupby("Ville")["Nom"].count()
print(r2)
```

## Exercice 8 : Pivot Table (Tables croisées dynamiques)

**Objectif :**
Créez une table croisée dynamique à partir de `df8` calculant la **moyenne des ventes** avec les **Régions** en lignes et les **Produits** en colonnes.

```python
df8 = pd.DataFrame({
    'Region': ['Nord', 'Nord', 'Sud', 'Sud', 'Nord', 'Sud'],
    'Produit': ['A', 'B', 'A', 'B', 'A', 'B'],
    'Ventes': [100, 200, 150, 250, 120, 220]
})

tableau = pd.pivot_table(
    df8,
    values="Ventes",
    index="Region",
    columns="Produit",
    aggfunc="mean"
)

print(tableau)
```

## Exercice 9 : Manipulation de séries temporelles

**Objectif :**
1. Créez un DataFrame avec une colonne de dates journalières sur le mois de Janvier 2026 et une colonne de ventes aléatoires.
2. Définissez la date comme index.
3. Rééchantillonnez les données pour obtenir la **somme hebdomadaire** (`'W'`).

```python
np.random.seed(42)

import pandas as pd
import numpy as np

dates = pd.date_range("2026-01-01", "2026-01-31", freq="D")

ventes = np.random.randint(100, 500, size=len(dates))

df = pd.DataFrame({
    "Date": dates,
    "Ventes": ventes
})

df = df.set_index("Date")

resultat = df["Ventes"].resample("W").sum()

print(resultat)
```

## Exercice 10 : Nettoyage et chaînage de méthodes (Method Chaining)

**Objectif :**
En une seule séquence d'instructions (chaînage) à partir de `df10` :
1. Convertissez les noms de villes en majuscules.
2. Filtrez pour ne garder que les lignes où `Ventes > 100`.
3. Ajoutez une colonne `'TVA'` (10% des ventes).

```python
df10 = pd.DataFrame({
    'Ville': ['paris', 'lyon', 'marseille', 'lille'],
    'Ventes': [80, 150, 200, 95]
})
df10 = (
    df10
    .assign(Ville=lambda df: df["Ville"].str.upper())
    .query("Ventes > 100")
    .assign(TVA=lambda df: df["Ventes"] * 0.10)
)

print(df10)

```
